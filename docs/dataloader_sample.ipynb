```python
import pandas as pd

pd.set_option("display.max_columns", None)
```

## Setup

```python
# setup your dataset path
dataset_local_path = "/tmp/"
y, m, d = "2024", "03", "01"
```

## AFC

```python
afc = pd.read_parquet(f"{dataset_local_path}afc-{y}-{m}-{d}.parquet")
afc.head()
```

## AVL

### AVL-lines

```python
avl_lines = pd.read_parquet(f"{dataset_local_path}avl-lines-{y}-{m}-{d}.parquet")
avl_lines.head()
```

### AVL-vehicles

```python
avl_veic = pd.read_parquet(f"{dataset_local_path}avl-vehicles-{y}-{m}-{d}.parquet")
avl_veic.head()
```

### AVL full

```python
# stop by stop in vehicle running
avl = avl_veic.merge(avl_lines, 
                     on=["route_short_name", "direction_id", "stop_id", "service_code", "longitude", "latitude"], 
                     how="left")
```

```python
avl.head()
```
